# Q&A Section

# a. When does the code in a function execute: when the function is defined or called?
# Answer: Code inside a function executes when the function is CALLED, not when it is defined.

# b. How many global scopes are there in a Python program? How many local scopes?
# Answer: There is only ONE global scope in a Python program.
# Each function call creates its own local scope, so there can be many local scopes.

# c. What happens to variables in a local scope when the function call returns?
# Answer: They are destroyed (garbage collected) and no longer exist outside the function.

# d. How can you prevent a program from crashing when it gets an error?
# Answer: Use try/except statements to catch errors and handle them gracefully.

```python
def collatz(number):
    if number % 2 == 0:
        result = number // 2
    else:
        result = 3 * number + 1
    print(result)
    return result
try:
    num = int(input("Enter number:\n"))
    while num != 1:
        num = collatz(num)
except ValueError:
    print("Invalid input! Please enter an integer.")
```
