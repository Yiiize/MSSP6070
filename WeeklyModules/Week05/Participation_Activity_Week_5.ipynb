```python
from google.colab import userdata
import os

github_token = userdata.get('Git_Key')
owner = 'Yiiize' # Replace with the GitHub repository owner
repository = 'MSSP6070' # Replace with the GitHub repository name

clone_url = f'https://{github_token}@github.com/{owner}/{repository}.git'

# Clone the repository
import subprocess
subprocess.run(['git', 'clone', clone_url])

# Navigate into the cloned repository directory (optional, but often useful)
os.chdir(repository)
print(f"Changed directory to: {os.getcwd()}")
```

```python
import pandas as pd
repository_dir = os.getcwd()
data_dir = os.path.join(repository_dir, 'data')

#b: read the file
filenames = ["aapl.csv", "amzn.csv", "fb.csv", "goog.csv", "nflx.csv"]
tickers = ["AAPL", "AMZN", "FB", "GOOG", "NFLX"]

dfs = {} # Dictionary to store dataframes

#c: add a ticker
for filename, ticker in zip(filenames, tickers):
    filepath = os.path.join(data_dir, filename)
    try:
        df = pd.read_csv(filepath)
        df['ticker'] = ticker
        dfs[ticker] = df # Store dataframe in dictionary
        print(f"DataFrame for {ticker}:")
        display(df.head())
    except FileNotFoundError:
        print(f"Error: The file {filepath} was not found.")
```

```python

# check dfs
if 'dfs' not in globals() or not dfs:
    raise ValueError("⚠️ 未找到 dfs，请先运行读取五个 CSV 的代码。")

#d: combine all DataFrame
faanf = pd.concat(dfs.values(), ignore_index=True)

# ensure the path
os.makedirs('data', exist_ok=True)

#e: save as csv doc
faanf.to_csv('data/faanf.csv', index=False)

print("✅ 已将 dfs 合并并保存为 data/faanf.csv")
print("文件行列数：", faanf.shape)

# see first rows
display(faanf.head())
```

```python

# f:read csv doc
df = pd.read_csv('data/faanf.csv')  # 如果文件名是 faanf.csv，请改成 'data/faanf.csv'

# colmns small letter
df.columns = [c.lower() for c in df.columns]

# change date to datetime
df['date'] = pd.to_datetime(df['date'], errors='coerce')
if df['date'].isna().any():
    print(f"⚠️ {df['date'].isna().sum()} date(s) could not be parsed and are NaT")

# change volume to int
df['volume'] = df['volume'].astype(str).str.replace(',', '', regex=False)  # 移除逗号
df['volume'] = pd.to_numeric(df['volume'], errors='coerce').fillna(0).astype(int)

# ensure ticker exist and big letters
if 'ticker' not in df.columns:
    raise KeyError("Ticker column not found")
df['ticker'] = df['ticker'].astype(str).str.upper()

# sort date and ticker
df = df.sort_values(by=['date','ticker']).reset_index(drop=True)

# review first rows
print("✅ 数据类型和前几行：")
print(df.dtypes)
display(df.head())

# save as csv
df.to_csv('data/fanng_cleaned.csv', index=False)
print("✅ 已保存 cleaned CSV 为 data/fanng_cleaned.csv")
```

```python

# g: Find the seven rows with the highest volume
top7 = df.nlargest(7, 'volume')  # sorts descending by volume automatically
print("Top 7 rows by volume:")
display(top7)

# h: Melt the DataFrame
# Keep 'date' and 'ticker' as identifiers, melt all other columns
id_vars = ['date', 'ticker']
value_vars = [col for col in df.columns if col not in id_vars]

melted = pd.melt(df, id_vars=id_vars, value_vars=value_vars,
                 var_name='variable', value_name='value')

print("Melted DataFrame preview:")
display(melted.head())

```
