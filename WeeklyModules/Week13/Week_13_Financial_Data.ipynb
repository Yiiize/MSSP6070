
In the context of downloading data from Yahoo Finance using the yfinance library, the period parameter defines the range of data to download and can have the following values:

*    1d, 5d: data for the past 1 or 5 days

*   1mo, 3mo, 6mo: data for the past 1, 3, or 6 months
*   1y, 2y, 5y, 10y, ytd: data for the past 1, 2, 5, or 10 years, or year to date.


*   max: data for the maximum available period.


These values allow you to specify the time frame for which you want to retrieve historical data for a given stock.

#Task 1: Read into seperate data frame the following tickers: Apple, Facebook, Google, and Netflix; use Yahoo finance.

```python
apple_df = yf.download('AAPL', period='max')
facebook_df = yf.download('META', period='max') # Facebook is now Meta Platforms, Inc.
google_df = yf.download('GOOGL', period='max') # Using GOOGL for Class A shares
netflix_df = yf.download('NFLX', period='max')

print("Apple Data (first 5 rows):")
print(apple_df.head())
print("\nFacebook/Meta Data (first 5 rows):")
print(facebook_df.head())
print("\nGoogle Data (first 5 rows):")
print(google_df.head())
print("\nNetflix Data (first 5 rows):")
print(netflix_df.head())
```

#Task 2:  Join the individual stock data frames into a single data frame.

```python
import pandas as pd

# Extract 'Close' price for each stock
apple_close = apple_df['Close']
facebook_close = facebook_df['Close']
google_close = google_df['Close']
netflix_close = netflix_df['Close']

# Rename the Series by assigning to the .name attribute
apple_close.name = 'AAPL_Close'
facebook_close.name = 'META_Close'
google_close.name = 'GOOGL_Close'
netflix_close.name = 'NFLX_Close'

# Join them into a single DataFrame
combined_df = pd.concat([apple_close, facebook_close, google_close, netflix_close], axis=1)

# Display the first few rows of the combined DataFrame
print("Combined Stock Data (first 5 rows):")
display(combined_df.head())
```

# Task 3: Create a visualization of the following financial metrics for each stock:  quickRatio, currentRatio, totalRevenue, debtToEquity, revenuePerShare, returnOnAssets, returnOnEquity, grossProfits


```python
import matplotlib.pyplot as plt
import seaborn as sns

# Define the tickers and the metrics to visualize
tickers = ['AAPL', 'META', 'GOOGL', 'NFLX']
metrics = [
    'quickRatio', 'currentRatio', 'totalRevenue', 'debtToEquity',
    'revenuePerShare', 'returnOnAssets', 'returnOnEquity', 'grossProfits'
]

# Create Ticker objects for each stock
apple_ticker = yf.Ticker('AAPL')
meta_ticker = yf.Ticker('META')
google_ticker = yf.Ticker('GOOGL')
netflix_ticker = yf.Ticker('NFLX')

# Dictionary to store the info for each ticker
ticker_infos = {
    'AAPL': apple_ticker.info,
    'META': meta_ticker.info,
    'GOOGL': google_ticker.info,
    'NFLX': netflix_ticker.info
}

# Extract the requested metrics for each stock
metrics_data = {}
for metric in metrics:
    metric_values = {}
    for ticker_symbol, info in ticker_infos.items():
        value = info.get(metric)
        metric_values[ticker_symbol] = value
    metrics_data[metric] = metric_values

# Convert to DataFrame for easier plotting
metrics_df = pd.DataFrame(metrics_data)

# Transpose the DataFrame to have stocks as index and metrics as columns
metrics_df_transposed = metrics_df.T
metrics_df_transposed.columns = tickers

# Plotting the financial metrics
plt.figure(figsize=(15, 20))
for i, metric in enumerate(metrics):
    plt.subplot(len(metrics), 1, i + 1) # Create a subplot for each metric
    sns.barplot(x='Ticker', y=metric, data=metrics_df_transposed.T.reset_index().rename(columns={'index': 'Ticker'}))
    plt.title(f'{metric} for Stocks')
    plt.ylabel(metric)
    plt.xlabel('')
plt.tight_layout()
plt.show()
```

# Task 5: Create a daily trading volume visualization for the year 2023 for each stock.


```python
import matplotlib.pyplot as plt
import pandas as pd

# Define the start and end dates for 2023
start_date_2023 = '2023-01-01'
end_date_2023 = '2023-12-31'

# Extract and filter volume data for each stock for 2023
apple_volume_2023 = apple_df.loc[start_date_2023:end_date_2023, 'Volume']
apple_volume_2023.name = 'AAPL_Volume'
facebook_volume_2023 = facebook_df.loc[start_date_2023:end_date_2023, 'Volume']
facebook_volume_2023.name = 'META_Volume'
google_volume_2023 = google_df.loc[start_date_2023:end_date_2023, 'Volume']
google_volume_2023.name = 'GOOGL_Volume'
netflix_volume_2023 = netflix_df.loc[start_date_2023:end_date_2023, 'Volume']
netflix_volume_2023.name = 'NFLX_Volume'

# Combine the volume data into a single DataFrame
combined_volume_2023 = pd.concat([
    apple_volume_2023,
    facebook_volume_2023,
    google_volume_2023,
    netflix_volume_2023
], axis=1)

# Plotting the daily trading volume for 2023
plt.figure(figsize=(15, 8))
combined_volume_2023.plot(ax=plt.gca())
plt.title('Daily Trading Volume for Apple, Meta, Google, and Netflix in 2023')
plt.xlabel('Date')
plt.ylabel('Volume')
plt.grid(True)
plt.legend(title='Stock')
plt.tight_layout()
plt.show()
```
