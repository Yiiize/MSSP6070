```python
# Step 1: Import pandas
import pandas as pd

# Step 2: Load the dataset
# (Replace the path with your uploaded file path in Colab)
df = pd.read_csv("/content/MSSP6070/data/Student Performance Data.csv")

# Step 3: Define a lambda function
# Example: Keep only math, reading, and writing scores, and add a new column 'average_score'
process_data = lambda data: data[['math score', 'reading score', 'writing score']].assign(
    average_score=lambda x: x.mean(axis=1)
)

# Step 4: Apply the lambda function to create a new DataFrame
new_df = process_data(df)

# Step 5: Display the result
new_df.head()
```
