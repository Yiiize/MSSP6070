```python
import pandas as pd
```

```python
df = pd.read_csv('parsed.csv')
```

```python
# a. Find the 95th percentile of earthquake magnitude in Japan using magType 'mb'
japan_mb = df[(df['parsed_place'].str.contains("Japan", na=False)) & (df['magType'] == 'mb')]
percentile_95_japan = japan_mb['mag'].quantile(0.95)
print("a. 95th percentile magnitude in Japan (mb):", percentile_95_japan)
```

```python
# b. Percentage of earthquakes in Indonesia coupled with tsunamis
indonesia = df[df['parsed_place'].str.contains("Indonesia", na=False)]
tsunami_percentage_indonesia = (indonesia['tsunami'].astype(int).sum() / len(indonesia)) * 100 if len(indonesia) > 0 else 0
print("b. % of Indonesia earthquakes with tsunami:", tsunami_percentage_indonesia)
```

```python
# c. Summary statistics for earthquakes in Nevada
nevada = df[df['parsed_place'].str.contains("Nevada", na=False)]
summary_stats_nevada = nevada['mag'].describe()
print("c. Summary statistics for Nevada earthquakes:\n", summary_stats_nevada)
```

```python
# d. Add a column for Ring of Fire locations
ring_of_fire_locations = [
    "Bolivia", "Chile", "Ecuador", "Peru", "Costa Rica", "Guatemala", "Mexico",
    "Japan", "Philippines", "Indonesia", "New Zealand", "Antarctica", "Canada",
    "Fiji", "Alaska", "Washington", "California", "Russia", "Taiwan", "Tonga",
    "Kermadec Islands"
]

df['Ring_of_Fire'] = df['parsed_place'].apply(
    lambda x: any(country in str(x) and "New Mexico" not in str(x) for country in ring_of_fire_locations)
)
print("d. Column 'Ring_of_Fire' added to dataframe")
```

```python
# e. Count earthquakes in and outside the Ring of Fire
ring_of_fire_count = df['Ring_of_Fire'].sum()
outside_ring_of_fire_count = len(df) - ring_of_fire_count
print("e. Counts - Ring of Fire:", ring_of_fire_count, ", Outside Ring of Fire:", outside_ring_of_fire_count)
```

```python

# f. Tsunami count along the Ring of Fire
tsunami_ring_of_fire_count = df[df['Ring_of_Fire']]['tsunami'].astype(int).sum()
print("f. Tsunami count along Ring of Fire:", tsunami_ring_of_fire_count)
```
