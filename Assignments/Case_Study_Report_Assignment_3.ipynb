```python
import pandas as pd

file_path = '/content/MSSP6070/data/HousingPrices.csv'
housing_data = pd.read_csv(file_path)

print("HousingPrices.csv loaded successfully.")
print(housing_data.head())
```

```python
mean_sale_price = housing_data['SalePrice'].mean()
std_sale_price = housing_data['SalePrice'].std()

print(f"Mean Sale Price: ${mean_sale_price:,.2f}")
print(f"Standard Deviation of Sale Price: ${std_sale_price:,.2f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(housing_data['SalePrice'], kde=True, bins=30)
plt.title('Distribution of Housing Sale Prices')
plt.xlabel('Sale Price ($)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.boxplot(y=housing_data['SalePrice'])
plt.title('Box Plot of Housing Sale Prices')
plt.ylabel('Sale Price ($)')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
