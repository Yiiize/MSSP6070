```python
# === Adventure Works Internet Sales — One-File Analysis Script ===
# How to use:
# 1) Put InternetSales.csv in your Colab working dir (or mount Drive)
# 2) Set DATA_PATH below and run all cells
# 3) Outputs will be saved to outputs/ ; report at outputs/Leadership_Report.md

import os, math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from datetime import datetime

# Optional: SciPy for Welch t-test; will fallback if not present
try:
    from scipy import stats
    SCIPY_AVAILABLE = True
except Exception:
    SCIPY_AVAILABLE = False

# -------- Paths --------
OUTPUT_DIR = Path("outputs")
OUTPUT_DIR.mkdir(exist_ok=True)
DATA_PATH = "/content/MSSP6070/data/InternetSales.csv"  # e.g. "/content/MyDrive/MSSP6070/data/InternetSales.csv"

# -------- Load --------
df = pd.read_csv(DATA_PATH,encoding="latin1")
print("Rows:", len(df), "Cols:", len(df.columns))
print("Columns:", list(df.columns)[:20])

# -------- Column mapping helper --------
def find_col(candidates, frame):
    cols = {c.lower(): c for c in frame.columns}
    for cand in candidates:
        if cand.lower() in cols:
            return cols[cand.lower()]
    # substring fallback
    for cand in candidates:
        for k_lower, orig in cols.items():
            if cand.lower() in k_lower:
                return orig
    return None

mapping = {
    "Year": find_col(["OrderYear","Year","Order Date","OrderDate","OrderDateKey"], df),
    "CountryRegion": find_col(["CountryRegion","CountryRegionName","SalesTerritoryCountry","Country"], df),
    "ProductCategory": find_col(["ProductCategory","EnglishProductName","Category"], df),
    "UnitPrice/Price": find_col(["UnitPrice","ListPrice","Price"], df),
    "SalesAmount": find_col(["SalesAmount","Sales","ExtendedAmount","LineTotal"], df),
    "YearlyIncome": find_col(["YearlyIncome","Income"], df),
    "Gender": find_col(["Gender"], df),
    "HomeOwner": find_col(["HomeOwnerFlag","HouseOwnerFlag","Homeowner","IsHomeOwner"], df),
}
print("Initial mapping:", mapping)

# -------- Cleaning / derivations --------
df_work = df.copy()

# Ensure Year as integer even if a date
if mapping["Year"] is None:
    date_col = find_col(["OrderDate","Order Date"], df_work)
    if date_col is not None:
        df_work["__Year"] = pd.to_datetime(df_work[date_col], errors="coerce").dt.year
        mapping["Year"] = "__Year"
else:
    # If looks like date string, parse to Year
    if df_work[mapping["Year"]].dtype == "object":
        maybe_date = pd.to_datetime(df_work[mapping["Year"]], errors="coerce")
        if maybe_date.notna().any():
            df_work["__Year"] = maybe_date.dt.year
            mapping["Year"] = "__Year"

# Normalize HomeOwner -> {0,1} if possible
if mapping["HomeOwner"] is not None:
    col = mapping["HomeOwner"]
    if df_work[col].dtype == "object":
        df_work["__HomeOwner"] = (
            df_work[col].astype(str).str.strip().str.lower().map({
                "y":1,"yes":1,"true":1,"1":1,"t":1,"owner":1,
                "n":0,"no":0,"false":0,"0":0,"f":0,"renter":0
            })
        )
        if df_work["__HomeOwner"].isna().mean() > 0.5:
            df_work["__HomeOwner"] = pd.to_numeric(df_work[col], errors="coerce")
    else:
        df_work["__HomeOwner"] = pd.to_numeric(df_work[col], errors="coerce")
    mapping["HomeOwner"] = "__HomeOwner"

print("Final mapping:", mapping)

# =========================
# Q1. Internet sales by Year × Country Region
# =========================
for k in ["Year","CountryRegion","SalesAmount"]:
    if mapping[k] is None:
        raise ValueError(f"Missing required column for {k}.")

sales_by_year_country = (
    df_work[[mapping["Year"], mapping["CountryRegion"], mapping["SalesAmount"]]]
    .dropna()
    .groupby([mapping["Year"], mapping["CountryRegion"]], dropna=False)
    .sum(numeric_only=True)
    .reset_index()
    .rename(columns={mapping["Year"]:"Year", mapping["CountryRegion"]:"CountryRegion", mapping["SalesAmount"]:"SalesAmount"})
)
pivot_sales = sales_by_year_country.pivot(index="Year", columns="CountryRegion", values="SalesAmount").fillna(0)
pivot_sales.to_csv(OUTPUT_DIR/"sales_by_year_country.csv", index=True)
print("\n[Q1] Sales by Year × CountryRegion (head):")
print(pivot_sales.head(10))

# Plot: total sales by year (all countries)
totals_by_year = sales_by_year_country.groupby("Year")["SalesAmount"].sum()
plt.figure()
totals_by_year.plot(kind="bar")
plt.title("Adventure Works: Total Internet Sales by Year")
plt.xlabel("Year")
plt.ylabel("Sales Amount")
plt.grid(axis="y")
plt.tight_layout()
# Save BEFORE show to avoid blank file in some backends
plt.savefig(OUTPUT_DIR/"sales_total_by_year.png")
plt.show()

# =========================
# Q2. Yearly Income stats (min, max, std) by Gender × Country Region
# =========================
for k in ["YearlyIncome","Gender","CountryRegion"]:
    if mapping[k] is None:
        raise ValueError(f"Missing required column for {k} (needed for income stats).")

g_stats = (
    df_work[[mapping["YearlyIncome"], mapping["Gender"], mapping["CountryRegion"]]]
    .dropna()
    .groupby([mapping["CountryRegion"], mapping["Gender"]])
    .agg(
        min_income=(mapping["YearlyIncome"], "min"),
        max_income=(mapping["YearlyIncome"], "max"),
        std_income=(mapping["YearlyIncome"], "std"),
        mean_income=(mapping["YearlyIncome"], "mean"),
        n=("Gender", "count"),
    )
    .reset_index()
    .rename(columns={mapping["CountryRegion"]:"CountryRegion", mapping["Gender"]:"Gender"})
)
g_stats.to_csv(OUTPUT_DIR/"income_stats_by_gender_country.csv", index=False)
print("\n[Q2] Income stats by Gender × CountryRegion (head):")
print(g_stats.head(10))

# =========================
# Q3. Average price for the top-sales product category
# =========================
for k in ["ProductCategory","SalesAmount"]:
    if mapping[k] is None:
        raise ValueError(f"Missing required column for {k} (needed for category sales).")

sales_by_cat = (
    df_work[[mapping["ProductCategory"], mapping["SalesAmount"]]]
    .dropna()
    .groupby(mapping["ProductCategory"])
    .sum(numeric_only=True)
    .sort_values(mapping["SalesAmount"], ascending=False)
)
top_category = sales_by_cat.index[0]
top_category_sales = float(sales_by_cat.iloc[0][mapping["SalesAmount"]])

if mapping["UnitPrice/Price"] is None:
    raise ValueError("Missing a price column (e.g., UnitPrice/ListPrice/Price).")

avg_price_top_cat = (
    df_work.loc[df_work[mapping["ProductCategory"]]==top_category, mapping["UnitPrice/Price"]]
    .astype(float)
    .dropna()
    .mean()
)

print("\n[Q3] Top Category:", top_category)
print("     Total Sales (Top Category):", round(top_category_sales, 2))
print("     Average Price (Top Category):", round(float(avg_price_top_cat), 2))

# Plot: sales by category
cat_totals = (
    df_work[[mapping["ProductCategory"], mapping["SalesAmount"]]]
    .dropna()
    .groupby(mapping["ProductCategory"])["SalesAmount"]
    .sum()
    .sort_values(ascending=False)
)
plt.figure()
cat_totals.plot(kind="bar")
plt.title("Internet Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Sales Amount")
plt.grid(axis="y")
plt.tight_layout()
plt.savefig(OUTPUT_DIR/"sales_by_category.png")
plt.show()

# =========================
# Q4. Does owning a home make a difference?
# =========================
for k in ["HomeOwner","SalesAmount"]:
    if mapping[k] is None:
        raise ValueError(f"Missing required column for {k} (needed for home ownership analysis).")

sub = df_work[[mapping["HomeOwner"], mapping["SalesAmount"]]].dropna()
sub = sub[(sub[mapping["HomeOwner"]].isin([0,1]))]

home_sales = sub.loc[sub[mapping["HomeOwner"]]==1, mapping["SalesAmount"]].astype(float)
rent_sales = sub.loc[sub[mapping["HomeOwner"]]==0, mapping["SalesAmount"]].astype(float)

def welch_ci(a, b, conf=0.95):
    ma, mb = a.mean(), b.mean()
    va, vb = a.var(ddof=1), b.var(ddof=1)
    na, nb = len(a), len(b)
    diff = ma - mb
    se = math.sqrt(va/na + vb/nb)
    # Welch–Satterthwaite df
    df = (va/na + vb/nb)**2 / ((va**2)/((na**2)*(na-1)) + (vb**2)/((nb**2)*(nb-1)))
    if SCIPY_AVAILABLE:
        from scipy import stats
        tcrit = stats.t.ppf((1+conf)/2, df)
    else:
        tcrit = 1.96
    lo, hi = diff - tcrit*se, diff + tcrit*se
    return diff, (lo, hi), df

diff, (lo, hi), df_w = welch_ci(home_sales, rent_sales)
if SCIPY_AVAILABLE:
    tstat, pval = stats.ttest_ind(home_sales, rent_sales, equal_var=False)
else:
    tstat = diff / math.sqrt(home_sales.var(ddof=1)/len(home_sales) + rent_sales.var(ddof=1)/len(rent_sales))
    # 2-sided normal approx
    pval = 2*(1 - 0.5*(1+math.erf(abs(tstat)/math.sqrt(2))))

summary_home = pd.DataFrame({
    "Group": ["HomeOwner=1","HomeOwner=0"],
    "N": [len(home_sales), len(rent_sales)],
    "MeanSales": [home_sales.mean(), rent_sales.mean()],
    "StdSales": [home_sales.std(ddof=1), rent_sales.std(ddof=1)]
})
summary_home.to_csv(OUTPUT_DIR/"homeowner_sales_summary.csv", index=False)
print("\n[Q4] Home Ownership summary:")
print(summary_home)
print(f"\nMean difference (Homeowners - Non-Homeowners): {diff:.2f}")
print(f"95% CI: [{lo:.2f}, {hi:.2f}]   df≈{df_w:.1f}")
print(f"Welch t-test: t={tstat:.3f}  p={pval:.4f}")

# Plot: mean sales by HomeOwner
means = sub.groupby(mapping["HomeOwner"])[mapping["SalesAmount"]].mean()
plt.figure()
means.plot(kind="bar")
plt.title("Mean Internet Sales by Home Ownership")
plt.xlabel("HomeOwner (0=No, 1=Yes)")
plt.ylabel("Mean Sales Amount")
plt.grid(axis="y")
plt.tight_layout()
plt.savefig(OUTPUT_DIR/"mean_sales_by_homeowner.png")
plt.show()

# =========================
# Export leadership report (Markdown) — ready to paste to Case Study Template
# =========================
report_path = OUTPUT_DIR/"Leadership_Report.md"

yr_examples = pivot_sales.head(10).to_markdown()
income_examples = g_stats.head(10).to_markdown(index=False)
top_cat_str = str(top_category)
avg_price_str = f"{avg_price_top_cat:.2f}"
home_tbl_md = summary_home.to_markdown(index=False)
today = datetime.now().strftime("%Y-%m-%d")

lines = []
lines.append(f"# Adventure Works Internet Sales — Leadership Report\n\nDate: {today}\n")
lines.append("## Executive Summary\n\n")
lines.append(
    "This analysis summarizes Internet sales by year and country region, explores demographic income patterns by "
    "gender and region, identifies the top-selling product category and its average price, and evaluates whether "
    "home ownership is associated with different spending behavior. Key charts and tables are included, with "
    "tactical recommendations to grow Internet sales.\n"
)
lines.append("\n## Data & Limitations\n\n")
lines.append(
    "Source: Microsoft AdventureWorks (fictitious). Data may include synthetic artifacts and missing values. "
    "Column names vary across exports; this analysis uses robust autodetection. All findings are for learning use only.\n"
)
lines.append("\n## Findings\n\n")
lines.append("### Internet Sales by Year × Country Region\n\n")
lines.append("The bar chart `sales_total_by_year.png` shows total sales by year; pivot excerpt below:\n\n")
lines.append(yr_examples + "\n")
lines.append("\n### Income Summary (YearlyIncome) by Country Region × Gender\n\n")
lines.append(income_examples + "\n")
lines.append("\n### Top Product Category\n\n")
lines.append(
    f"The highest-grossing product category is **{top_cat_str}**. Its average product price is **{avg_price_str}**. "
    "See `sales_by_category.png` for category sales distribution.\n"
)
lines.append("\n### Home Ownership & Sales\n\n")
lines.append("We compare mean sales between homeowners and non-homeowners (Welch two-sample t-test).\n\n")
lines.append(home_tbl_md + "\n\n")
lines.append(f"Mean difference (Homeowners - Non-Homeowners): **{diff:.2f}**; 95% CI [{lo:.2f}, {hi:.2f}]; p={pval:.4f}.\n")
lines.append("\n## Recommendations\n\n")
lines.append(
    f"1. **Double down on {top_cat_str}**: build bundles and accessory upsells around the {avg_price_str} anchor.\n"
    "2. **Region-focused campaigns**: prioritize countries/years with strongest historical demand; localize promos/shipping.\n"
    "3. **Price packaging**: test good–better–best tiers and cart cross-sells tied to category leaders.\n"
    "4. **Segment messaging**: if homeowner spending differs, tailor financing/options and creative.\n"
    "5. **Funnel hygiene**: speed up PDP/checkout, reduce abandonment, and retarget with simple time-boxed offers.\n"
)
lines.append("\n## APA/Presentation Checklist\n\n")
lines.append(
    "- One chart per figure; labeled axes/titles; default colors.\n"
    "- Tables exported with clear headers; formatting applied in the final doc.\n"
    "- Explicitly state dataset limitations.\n"
)
report_path.write_text("".join(lines), encoding="utf-8")
print("\nWrote report to:", report_path.resolve())
print("Artifacts saved to:", OUTPUT_DIR.resolve())
```
